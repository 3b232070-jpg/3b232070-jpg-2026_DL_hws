```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import confusion_matrix
import seaborn as sns
import ipywidgets as widgets

# ==========================================
# 1. 資料預處理
# ==========================================
(X_train, y_train), (X_test, y_test) = cifar10.load_data()

class_names = [
    'airplane', 'automobile', 'bird', 'cat', 'deer',
    'dog', 'frog', 'horse', 'ship', 'truck'
]

print("訓練資料 X_train 形狀:", X_train.shape)
print("訓練標籤 y_train 形狀:", y_train.shape)
print("測試資料 X_test 形狀:", X_test.shape)
print("測試標籤 y_test 形狀:", y_test.shape)

# 呈現三張訓練圖片及其真實標籤名稱
plt.figure(figsize=(9, 3))
for i in range(3):
    plt.subplot(1, 3, i + 1)
    plt.imshow(X_train[i])
    plt.title(f"Label: {class_names[y_train[i][0]]}")
    plt.axis('off')
plt.show()

# 正規化與 One-hot Encoding
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
y_train_onehot = to_categorical(y_train, 10)
y_test_onehot = to_categorical(y_test, 10)

# ==========================================
# 2. 建立神經網路模型 (HW4-3 深層架構規格)
# ==========================================
model = Sequential([
    # 第 01~04 層
    Conv2D(32, (3, 3), padding='same', input_shape=(32, 32, 3), activation='relu'),
    Dropout(0.3),
    Conv2D(32, (3, 3), padding='same', activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),

    # 第 05~08 層
    Conv2D(64, (3, 3), padding='same', activation='relu'),
    Dropout(0.3),
    Conv2D(64, (3, 3), padding='same', activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),

    # 第 09~12 層
    Conv2D(128, (3, 3), padding='same', activation='relu'),
    Dropout(0.3),
    Conv2D(128, (3, 3), padding='same', activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),

    # 第 13~18 層
    Flatten(),
    Dropout(0.3),
    Dense(2500, activation='relu'),
    Dropout(0.3),
    Dense(1500, activation='relu'),
    Dropout(0.3),

    # 第 19 層 (輸出層)
    Dense(10, activation='softmax')
])

model.summary()

# ==========================================
# 3. 訓練神經網路模型 (epochs=50)
# ==========================================
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

history = model.fit(
    X_train, y_train_onehot,
    batch_size=128,
    epochs=50,
    validation_split=0.2,
    verbose=1
)

# ==========================================
# 4. 顯示訓練過程的 Loss & Accuracy
# ==========================================
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Val Accuracy')
plt.title('HW4-3 Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('HW4-3 Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# ==========================================
# 5. 評估模型準確率
# ==========================================
test_loss, test_acc = model.evaluate(X_test, y_test_onehot, verbose=0)
print(f"HW4-3 測試資料集準確率: {test_acc * 100:.2f}%")

# ==========================================
# 6. 進行預測與展示
# ==========================================
predictions = model.predict(X_test)
predicted_classes = np.argmax(predictions, axis=1)

plt.figure(figsize=(12, 5))
for i in range(5):
    idx = np.random.randint(0, len(X_test))
    plt.subplot(1, 5, i + 1)
    plt.imshow(X_test[idx])
    pred_name = class_names[predicted_classes[idx]]
    true_name = class_names[y_test[idx][0]]
    color = 'green' if predicted_classes[idx] == y_test[idx][0] else 'red'
    plt.title(f"Pred: {pred_name}\nTrue: {true_name}", color=color)
    plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# 7. 混淆矩陣 (Confusion Matrix)
# ==========================================
cm = confusion_matrix(y_test, predicted_classes)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('HW4-3 Confusion Matrix')
plt.show()

# ==========================================
# 8. 儲存完整模型 (.keras 格式)
# ==========================================
model.save("hw4_3_model.keras")
print("模型已成功儲存為 hw4_3_model.keras")

# ==========================================
# 9. interact_manual 互動預測
# ==========================================
def predict_test_image(index):
    img = X_test[index]
    plt.figure(figsize=(3, 3))
    plt.imshow(img)
    plt.axis('off')
    plt.show()

    pred_prob = model.predict(np.expand_dims(img, axis=0), verbose=0)
    pred_idx = np.argmax(pred_prob)
    print(f"真實標籤: {class_names[y_test[index][0]]}")
    print(f"預測結果: {class_names[pred_idx]} (信心機率: {pred_prob[0][pred_idx]*100:.2f}%)")

widgets.interact_manual(predict_test_image, index=widgets.IntSlider(min=0, max=999, step=1, value=0))
```
